```python
import json

print(json.dumps({"name": "John", "age": 30}))
print(json.dumps(["apple", "bananas"]))
print(json.dumps(("apple", "bananas")))
print(json.dumps("hello"))
print(json.dumps(42))
print(json.dumps(31.76))
print(json.dumps(True))
print(json.dumps(False))
print(json.dumps(None))
```

```python
#regular expression
txt='i am getting job soon, job will help me grow and bring success'
import re
x=re.findall("job",txt)
print(type(x),x)
y=re.search("job",txt)
print(type(y),y)
z=re.split("\\s",txt)
print(type(z),z)
t = re.sub("\\s", "9", txt)
print(t)
t_2 = re.sub("9", "4", t, 2) #Replace the first 2 occurrences:
print(t_2)
```
